### Monthly Data Coverage by Country

Examine the historical availability of monthly Battery Electric Vehicle (BEV)
registrations across European countries.

The objective is to identify differences in historical coverage and detect
missing months within each country's available observation period.


```python
# Convert Date into datetime format
df_ev_sales["Date"] = pd.to_datetime(df_ev_sales["Date"])

# Filter European countries and individual BEV observations
bev_europe = df_ev_sales[
    (df_ev_sales["Area type"] == "Country or economy")
    & (df_ev_sales["Continent"] == "Europe")
    & (df_ev_sales["Drivetrain"] == "Battery electric")
].copy()

# Calculate monthly coverage by country
country_coverage = (
    bev_europe.groupby("Area")
    .agg(
        first_month=("Date", "min"),
        last_month=("Date", "max"),
        available_months=("Date", "nunique"),
    )
    .reset_index()
)

# Calculate the expected number of months
country_coverage["expected_months"] = (
    (country_coverage["last_month"].dt.year - country_coverage["first_month"].dt.year)
    * 12
    + country_coverage["last_month"].dt.month
    - country_coverage["first_month"].dt.month
    + 1
)

# Identify missing months within each country's observation period
country_coverage["missing_months"] = (
    country_coverage["expected_months"] - country_coverage["available_months"]
)

# Sort countries by historical coverage
country_coverage = country_coverage.sort_values("first_month")

# Display the results
display(country_coverage)

# Identify countries with missing months
display(country_coverage[country_coverage["missing_months"] > 0])
```

#### Monthly Data Coverage

The dataset contains monthly BEV registration observations for 33 European
countries, with historical coverage varying by country.

Most countries have continuous monthly observations within their available
periods. Bulgaria has one missing month, which will be investigated separately.

The differences in historical coverage must be considered when comparing
countries or conducting time-series analysis.

A common time period may be selected for specific comparative analyses to
ensure consistency across countries.
