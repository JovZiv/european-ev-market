# Economic Indicators – Data Exploration

This notebook explores economic indicators across European countries, including GDP per capita, population, and electricity prices.

The objective is to assess data coverage, quality, and suitability for investigating potential factors influencing electric vehicle adoption.

```python
# Import libraries
import pandas as pd

# Load GDP per capita dataset
df_gdp = pd.read_csv("../data/raw/eurostat_gdp_per_capita.csv")

# Inspect dataset structure
print("Dataset shape:", df_gdp.shape)
display(df_gdp.head())
df_gdp.info()
```

### Data Coverage and Quality

The dataset contains annual GDP per capita values measured in chain-linked 2020 euros per capita.

The following checks assess geographic and temporal coverage, missing values, and duplicate country-year observations.

```python
# Check geographic and temporal coverage
print("Number of geographic entities:", df_gdp["geo"].nunique())
print("Countries / entities:", sorted(df_gdp["geo"].unique()))

print("Year range:", df_gdp["TIME_PERIOD"].min(), "-", df_gdp["TIME_PERIOD"].max())

# Check missing values
display(df_gdp.isna().sum())

# Check duplicate country-year combinations
duplicates = df_gdp.duplicated(subset=["geo", "TIME_PERIOD"]).sum()
print("Duplicate country-year combinations:", duplicates)

# Check unique country-year combinations
print(
    "Unique country-year combinations:",
    df_gdp[["geo", "TIME_PERIOD"]].drop_duplicates().shape[0],
)
```

### GDP per Capita – Value Checks

```python
# Check GDP per capita values
print("Negative values:", (df_gdp["OBS_VALUE"] < 0).sum())
print("Zero values:", (df_gdp["OBS_VALUE"] == 0).sum())

# Summary statistics
display(df_gdp["OBS_VALUE"].describe())

# Inspect the minimum and maximum observations
display(df_gdp.loc[df_gdp["OBS_VALUE"].idxmin(), ["geo", "TIME_PERIOD", "OBS_VALUE"]])

display(df_gdp.loc[df_gdp["OBS_VALUE"].idxmax(), ["geo", "TIME_PERIOD", "OBS_VALUE"]])
```

No negative or zero GDP per capita values were identified.

GDP per capita ranges from €2,880 in Albania (2005) to €109,090 in Luxembourg (2007). The maximum is high but plausible given differences in economic output across countries.

No observations were removed based on these checks.

# Population - Data Exploration

```python
# Load population dataset
df_population = pd.read_csv("../data/raw/eurostat_population.csv")

# Inspect dataset structure
print("Dataset shape:", df_population.shape)
display(df_population.head())
df_population.info()
```

### Population – Data Coverage and Quality

The dataset contains annual population estimates measured on 1 January.

```python
# Check geographic and temporal coverage
print("Number of geographic entities:", df_population["geo"].nunique())
print("Countries / entities:", sorted(df_population["geo"].unique()))

print(
    "Year range:",
    df_population["TIME_PERIOD"].min(),
    "-",
    df_population["TIME_PERIOD"].max(),
)

# Check missing values
display(df_population.isna().sum())

# Check duplicate country-year combinations
duplicates = df_population.duplicated(subset=["geo", "TIME_PERIOD"]).sum()

print("Duplicate country-year combinations:", duplicates)
```

### Population – Data Coverage and Quality

The dataset contains annual population estimates for 51 geographic entities covering 2015–2026.

No missing values were found in the main analytical columns, and there are no duplicate country-year observations.

The dataset includes European countries as well as EU and Euro area aggregates.
Aggregates will be excluded from country-level analysis.

Population is measured on 1 January and will be aligned with the corresponding year when integrating the dataset with EV adoption data.

```python
# Check population values
print("Negative values:", (df_population["OBS_VALUE"] < 0).sum())
print("Zero values:", (df_population["OBS_VALUE"] == 0).sum())

# Summary statistics
display(df_population["OBS_VALUE"].describe())

# Inspect minimum and maximum observations
display(
    df_population.loc[
        df_population["OBS_VALUE"].idxmin(), ["geo", "TIME_PERIOD", "OBS_VALUE"]
    ]
)

display(
    df_population.loc[
        df_population["OBS_VALUE"].idxmax(), ["geo", "TIME_PERIOD", "OBS_VALUE"]
    ]
)
```

### Population – Value Checks

No negative or zero population values were identified.

Population ranges from 32,789 in San Marino (2015) to 451,990,314 in the EU-27 aggregate (2026).

The maximum reflects an aggregate rather than an individual country. EU and Euro area aggregates will be excluded from country-level analysis.

No observations were removed based on these checks.

# Electricity Prices - Data Exploration

This section explores electricity prices across European countries.

Electricity prices may influence the operating cost of electric vehicles and could help explain differences in EV adoption.

The dataset is assessed for geographic and temporal coverage, data quality, and suitability for comparison with EV adoption.

```python
# Load electricity prices dataset
df_electricity = pd.read_csv("../data/raw/eurostat_electricity_prices.csv")

# Inspect dataset structure
print("Dataset shape:", df_electricity.shape)
display(df_electricity.head())
df_electricity.info()
```

```python
# Inspect the different electricity price categories
for column in ["freq", "siec", "nrg_cons", "unit", "tax", "currency"]:
    print(f"\n{column}:")
    display(df_electricity[column].value_counts())
```

```python
# Check geographic coverage
print("Number of geographic entities:", df_electricity["geo"].nunique())
print("Countries / entities:", sorted(df_electricity["geo"].unique()))

# Check temporal coverage
print("Periods:", sorted(df_electricity["TIME_PERIOD"].unique())[:5], "...")
print("First period:", df_electricity["TIME_PERIOD"].min())
print("Last period:", df_electricity["TIME_PERIOD"].max())

# Check missing values
display(df_electricity.isna().sum())

# Check duplicate country-period combinations
duplicates = df_electricity.duplicated(subset=["geo", "TIME_PERIOD"]).sum()

print("Duplicate country-period combinations:", duplicates)
```

### Electricity Prices – Data Coverage and Quality

The dataset contains half-yearly electricity prices for 43 geographic entities from 2007-S1 to 2026-S1.

No missing values were found in the main analytical columns, and there are no duplicate country-period observations.

The dataset includes EU and Euro area aggregates, which will be excluded from country-level analysis.

Electricity prices will be aggregated to annual values during data integration to align with the annual economic indicators.

```python
# Check electricity price values
print("Negative values:", (df_electricity["OBS_VALUE"] < 0).sum())
print("Zero values:", (df_electricity["OBS_VALUE"] == 0).sum())

# Summary statistics
display(df_electricity["OBS_VALUE"].describe())

# Inspect minimum and maximum observations
display(
    df_electricity.loc[
        df_electricity["OBS_VALUE"].idxmin(), ["geo", "TIME_PERIOD", "OBS_VALUE"]
    ]
)

display(
    df_electricity.loc[
        df_electricity["OBS_VALUE"].idxmax(), ["geo", "TIME_PERIOD", "OBS_VALUE"]
    ]
)
```

### Electricity Prices – Value Checks

No negative or zero electricity price values were identified.

Prices range from €0.0249/kWh in Ukraine (2016-S1) to €0.5871/kWh in Denmark (2022-S2).

The variation reflects differences across countries and time periods.
No observations were removed based on these checks.
